```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Done J")
```

```python
data = pd.read_csv("../../student_scores.csv")
data.head()
```

```python
print("Dataset shape:", data.shape)
print("\nBasic statistics:")
display(data.describe())

print("\nMissing values:")
display(data.isnull().sum())
```

```python
plt.figure(figsize=(8, 6))

plt.hist(
    data["Study Hours"],
    bins=5,
    color="teal",
    edgecolor="black"
)

plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.bar(
    data["Student"],
    data["Test Score"],
    color="brown",
    edgecolor="black"
)

plt.title("Test Scores of Students")
plt.xlabel("Student")
plt.ylabel("Test Score")
plt.xticks(data["Student"])
plt.ylim(0, 100)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.plot(
    data["Study Hours"],
    data["Test Score"],
    color="brown",
    marker="o",
    linewidth=2
)

plt.title("Study Hours vs Test Score")
plt.xlabel("Study Hours")
plt.ylabel("Test Score")
plt.ylim(0, 100)
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=data,
    x="Study Hours",
    y="Test Score",
    color="red",
    s=100
)

plt.title("Study Hours vs Test Score")
plt.xlabel("Study Hours")
plt.ylabel("Test Score")
plt.ylim(0, 100)
plt.show()

correlation = data["Study Hours"].corr(data["Test Score"])
print(f"Correlation: {correlation:.3f}")
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=data,
    x="Study Hours",
    y="Test Score",
    color="red",
    s=100
)

plt.title("Study Hours vs Test Score")
plt.xlabel("Study Hours")
plt.ylabel("Test Score")
plt.ylim(0, 100)
plt.show()

correlation = data["Study Hours"].corr(data["Test Score"])
print(f"Correlation: {correlation:.3f}")
```

```python
import numpy as np

# Prepare the variables
x = data["Study Hours"].to_numpy()
y = data["Test Score"].to_numpy()

# Fit a straight-line model: predicted score = slope × hours + intercept
slope, intercept = np.polyfit(x, y, 1)
predicted_scores = slope * x + intercept

# Calculate R-squared
ss_residual = np.sum((y - predicted_scores) ** 2)
ss_total = np.sum((y - np.mean(y)) ** 2)
r_squared = 1 - (ss_residual / ss_total)

print(f"Regression equation: Test Score = {slope:.2f} × Study Hours + {intercept:.2f}")
print(f"R-squared: {r_squared:.3f}")

# Plot observations and regression line
plt.figure(figsize=(8, 6))

sns.scatterplot(
    x=x,
    y=y,
    color="red",
    s=100,
    label="Actual scores"
)

plt.plot(
    x,
    predicted_scores,
    color="navy",
    linewidth=2,
    label="Regression line"
)

plt.title("Linear Regression: Study Hours and Test Score")
plt.xlabel("Study Hours")
plt.ylabel("Test Score")
plt.ylim(0, 100)
plt.legend()
plt.show()
```

```python
study_hours = 5.5
predicted_score = slope * study_hours + intercept

print(
    f"The model predicts a score of {predicted_score:.1f} "
    f"for a student who studies {study_hours} hours."
)
```
